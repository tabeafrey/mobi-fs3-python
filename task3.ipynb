## Linear filtering by convolution (mean filter)

**<a id="#">Task 3.1.1.</a>** Load the image `data/astronaut.png` and show it:

```python
img= plt.imread("data/astronaut.png")
plt.figure(figsize=(15,8))
plt.imshow(img, "gray")
plt.colorbar()
```

**<a id="#">Task 3.1.2.</a>** Implement a mean filter:

```python
def meanfilter(img_in, size):
    result = numpy.zeros(img_in.shape)
    n = (size - 1) // 2
    for p in numpy.ndindex(img_in.shape):
        neighborhood_sum = 0

        if (p[0] >= n and p[0] < img_in.shape[0]-n and
            p[1] >= n and p[1] < img_in.shape[1]-n):
            
            
            for q in numpy.ndindex(size, size):
                neighborhood_sum += img_in[
                    p[0] + q[0] - n,
                    p[1] + q[1] - n
                    ]
            mean= neighborhood_sum /(size * size)
            result[p[0], p[1]]= mean
            
    return result
```

**<a id="#">Task 3.1.3.</a>** Test your implementation from above by using the function `meanfilter` for the previously loaded image and show the result:

```python
meanf_img = meanfilter(img, 3)
plt.figure(figsize=(15,8))
plt.imshow(meanf_img, "gray")
plt.colorbar
```

**<a id="#">Task 3.1.4.</a>** Compare your result for filter size `5` with the correct result image `data/astronaut_meanfilter5.png`:

```python
meanf5_2 = plt.imread("data/astronaut_meanfilter5.png")
meanf5_img = meanfilter(img, 5)
plt.figure(figsize=(15,8))
plt.imshow(meanf5_img, "gray")
plt.colorbar
plt.figure(figsize=(15,8))
plt.imshow(meanf5_2, "gray")
plt.colorbar()
assert numpy.allclose (meanf5_2, meanf5_img, atol=1/255)
```

## Non-linear filtering

**<a id="#">Task 3.2.1.</a>** Implement a median filter:

```python
def medianfilter(img_in, size):
    result = numpy.zeros(img_in.shape)
    n = (size - 1) // 2
    for p in numpy.ndindex(img_in.shape):
        medianlist = list ()

        if (p[0] >= n and p[0] < img_in.shape[0]-n and
            p[1] >= n and p[1] < img_in.shape[1]-n):
            for q in numpy.ndindex(size, size):
                medianlist.append (img_in[
                    p[0] + q[0] - n,
                    p[1] + q[1] - n
                    ])
            medianlist.sort()
            result[p] = medianlist[(size * size)// 2]
            
    return result


   
```

**<a id="#">Task 3.2.2.</a>** First, test your implementation from above by using the function `medianfilter` for the previously loaded image and show the result:

```python
medianp = medianfilter(img, 5)
plt.figure(figsize=(15,8))
plt.imshow(medianp, "gray")
plt.colorbar
```

Then, compare your result for filter size `5` with the correct result image `data/astronaut_medianfilter5.png`:

```python
medianp2 = plt.imread("data/astronaut_medianfilter5.png")
plt.figure(figsize=(15,8))
plt.imshow(medianp2, "gray")
plt.colorbar
plt.figure(figsize=(15,8))
plt.imshow(medianp, "gray")
plt.colorbar()
assert numpy.allclose (medianp2, medianp, atol=1/255)
```
